```python
# Given an array of integers and a target value, return the indices of the two numbers whose sum equals the target.
# nums = [2, 7, 11, 15]
# target = 9

# A Hash Map is a data structure that stores data as key-value pairs.
#  It uses a hash function to map each key to a memory location, allowing data to be inserted, searched, and updated in O(1) average time.


def two_sum(nums, target):
    seen = {}
    for i in range(len(nums)):
        rem = target - nums[i]
        
        if rem in seen:
            return [seen[rem],i]
        
        else:
            seen[nums[i]] = i
            
    return []
    
nums = [2, 7, 11, 15]
target = 18

t = two_sum(nums,target)

print(t)
        
```

```python
# 2. Best Time to Buy and Sell Stock ⭐⭐⭐⭐



def max_profit(prices):
    profit = 0
    for i in range(len(prices)):
        for j in range(i+1,len(prices)):
            amt = prices[j] - prices[i]
            if amt > profit:
                profit = amt
    return profit
                
                
prices = [7,1,5,3,6,4]

a = max_profit(prices)

print(a)
        
```

```python
# Given N and K, traverse N till 0, and upon reaching 0 traverse it back to N using the difference of K.
# Note: Do not use any loops.”

def traverse(n, k):
    if n < 0:
        return
    print(n)
    traverse(n - k, k)
    if n != 0:
        print(n)


    
```

```python
# Because every recursive call (traverse(6,2), traverse(4,2), traverse(2,2)) is paused, not deleted. When traverse(0,2) returns, Python resumes the previous paused call and executes the line after the recursive call (print(n)).

def tran(N,K):
    print(N,end = " ")
    
    if N == 0:
        return
    
    tran(N-K, K)
    print(N,end=" ")
    
a = tran(12,3)
print(a)
```

```python
def tran(N,K):
    print(N,end = " ")
    
    if N == 0:
        return
    
    if N-K < 0:
        tran(0,K)
        
    else:
        tran(N-K, K)
    
    print(N,end=" ")

    
tran(10,3)
```
