```python
🧩 Types of Sorting Algorithms in DSA

There are two main categories of sorting algorithms:

Category	Example
Comparison-based sorting	Bubble Sort, Selection Sort, Insertion Sort, Merge Sort, Quick Sort, Heap Sort
Non-comparison-based sorting	Counting Sort, Radix Sort, Bucket Sort
```

```python
# 🧮 1️⃣ Bubble Sort
# 🔹 Idea:

# Compare adjacent elements, and swap them if they’re in the wrong order.
# Repeat this process until the array is sorted.



# [5, 2, 9, 1]
# Pass 1: [2, 5, 1, 9]
# Pass 2: [2, 1, 5, 9]
# Pass 3: [1, 2, 5, 9]
```

```python
# 🧮 2️⃣ Selection Sort
# 🔹 Idea:

# Select the smallest element in each pass and put it in its correct position.

# [5, 2, 9, 1]
# Find min (1), put at index 0 → [1, 2, 9, 5]
# Find next min (2) → already correct
# Find next min (5) → [1, 2, 5, 9]
```

```python
# 🧮 3️⃣ Insertion Sort : 
# Build the sorted array one element at a time.
# Pick each element and insert it into its correct position among the previously sorted elements.

[5, 2, 9, 1]
Start with [5]
Insert 2 → [2, 5]
Insert 9 → [2, 5, 9]
Insert 1 → [1, 2, 5, 9]
```

```python
| **Algorithm**         | **Pass 1**                                    | **Pass 2**                                       | **Pass 3**                                | **Final**        |
| --------------------- | --------------------------------------------- | ------------------------------------------------ | ----------------------------------------- | ---------------- |
| 🫧 **Bubble Sort**    | Compare and swap adjacent → `[3, 4, 1, 5]`    | Compare (3,4),(4,1) → `[3, 1, 4, 5]`             | Compare (3,1) → `[1, 3, 4, 5]`            | ✅ `[1, 3, 4, 5]` |
| 🎯 **Selection Sort** | Smallest (1) → swap with 1st → `[1, 3, 4, 5]` | (2nd min = 3) no swap → `[1, 3, 4, 5]`           | (3rd min = 4) no swap                     | ✅ `[1, 3, 4, 5]` |
| 🃏 **Insertion Sort** | Take 3 → insert before 5 → `[3, 5, 4, 1]`     | Take 4 → insert between 3 and 5 → `[3, 4, 5, 1]` | Take 1 → insert at start → `[1, 3, 4, 5]` | ✅ `[1, 3, 4, 5]` |


| **Algorithm**      | **Best Case**      | **Average Case** | **Worst Case** | **Space** | **Stable** | **Adaptive** |
| ------------------ | ------------------ | ---------------- | -------------- | --------- | ---------- | ------------ |
| **Bubble Sort**    | O(n) *(optimized)* | O(n²)            | O(n²)          | O(1)      | ✅ Yes      | ✅ Yes        |
| **Selection Sort** | O(n²)              | O(n²)            | O(n²)          | O(1)      | ❌ No       | ❌ No         |
| **Insertion Sort** | O(n)               | O(n²)            | O(n²)          | O(1)      | ✅ Yes      | ✅ Yes        |


```

```python
def search(nums, target):
    left = 0
    right = len(nums) - 1

    while left <= right:
        mid = (left + right) // 2

        if nums[mid] == target:
            return mid

        # Left half is sorted
        if nums[left] <= nums[mid]:

            if nums[left] <= target < nums[mid]:
                right = mid - 1
            else:
                left = mid + 1

        # Right half is sorted
        else:

            if nums[mid] < target <= nums[right]:
                left = mid + 1
            else:
                right = mid - 1

    return -1

a = [1, 2, 3, 4, 5, 6, 7]

c = search(a, 5)

print(c)  # Output: 4
```
