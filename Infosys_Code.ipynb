```python
# When a child class overrides a method or constructor of the parent class, the parent’s logic is hidden.
# super() allows you to reuse parent behavior instead of rewriting it.

class Parent:
    def greet(self):
        print("Hello from Parent")

class Child(Parent):
    def greet(self):
        super().greet()   # call parent method
        print("Hello from Child")

obj = Child()
obj.greet()
```

```python
class Bank:
  def __init__(self,balance):
    self.__balance = balance

  def deposit(self,amt):
    if amt>0:
      self.__balance += amt

  def withdraw(self,amt):
    if 0 < amt <= self.__balance:
      self.__balance -= amt

  def get_balance(self):
    return self.__balance


obj1 = Bank(1000)

obj1.deposit(500)

obj1.withdraw(52)

print(obj1.get_balance())
    

    
```

```python
# 📌 Interview line:Abstraction is about design, encapsulation is about data protection.

# | Abstraction          | Encapsulation         |
# | -------------------- | --------------------- |
# | Hides implementation | Hides data            |
# | WHAT system does     | HOW data is protected |
# | Design-level         | Code-level            |

# 🔥 Interview Trap Questions

# ❓ Can abstract class have normal methods?
# 👉 ✔ Yes

# ❓ Can we create object of abstract class?
# 👉 ❌ No

# ❓ Is abstraction mandatory?
# 👉 ❌ No, but recommended for large systems


from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using Credit Card")
```

```python
# Polymorphism allows the same method to perform different actions based on the object type.
# 🔹 Types of Polymorphism in Python
# 1️⃣ Compile-time Polymorphism

# ❌ Python does NOT support it directly
# (Java supports method overloading)

# 2️⃣ Runtime Polymorphism (MOST IMPORTANT)

# Achieved using:

# Method overriding
# Inheritance


class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")

obj1 = [Dog(), Cat()]

for a in obj1:
    a.sound()  # Runtime polymorphism
```

```python
# Polymorphism --> Duck Typing

class Dog:
    def sound(self):
        print("Dog barks")

class Cat:
    def sound(self):
        print("Cat meows")

# The function doesn't care whether animal is a Dog or Cat.

# It simply says:

# "Give me an object that has a speak() method."
def make_sound(animal):  
    animal.sound()

make_sound(Dog())  # Dog barks
make_sound(Cat())  # Cat meows
```

```python
# if we not required index then use set

def two_sum(nums, target):
    seen = set()

    for i in range(len(nums)):
        rem = target - nums[i]

        if rem in seen:
            return [rem, nums[i]]

        seen.add(nums[i])

    return []

nums = [2, 7, 11, 15]
target = 18

print(two_sum(nums, target))
```

```python
# if we required index then use dictionary

def two_sum(nums, target):
    seen = {}

    for i in range(len(nums)):
        rem = target - nums[i]

        if rem in seen:
            return [seen[rem], i]

        seen[nums[i]] = i

    return []

nums = [2, 7, 11, 15]
target = 18

print(two_sum(nums, target))
```

```python
s = "abcabcbb"

left = 0
max_length = 0
seen = set()

for right in range(len(s)):
    while s[right] in seen:
        seen.remove(s[left])
        left += 1

    seen.add(s[right])

    max_length = max(max_length, right - left + 1)

print(max_length)
```
