```python
# When a child class overrides a method or constructor of the parent class, the parent’s logic is hidden.
# super() allows you to reuse parent behavior instead of rewriting it.

class Parent:
    def greet(self):
        print("Hello from Parent")

class Child(Parent):
    def greet(self):
        super().greet()   # call parent method
        print("Hello from Child")

obj = Child()
obj.greet()
```

```python
class Bank:
  def __init__(self,balance):
    self.__balance = balance

  def deposit(self,amt):
    if amt>0:
      self.__balance += amt

  def withdraw(self,amt):
    if 0 < amt <= self.__balance:
      self.__balance -= amt

  def get_balance(self):
    return self.__balance


obj1 = Bank(1000)

obj1.deposit(500)

obj1.withdraw(52)

print(obj1.get_balance())
    

    
```
